```python
##Task 1.1 — Read the file the foundations way
import csv
import json


with open("data_appointments.csv", "r", newline="") as file:
    reader = csv.DictReader(file)

    print("Headers:", reader.fieldnames)

    rows = list(reader)

    print("Total rows:", len(rows))
```

```python
def parse_row(row):

    row['age']=int(row['age'])  # Convert age

    row['fee_inr']=int(row['fee_inr'])# Convert fee_inr

    row['prior_no_shows']=int(row['prior_no_shows'])# Convert prior_no_shows

    row['distance_km']=float(row['distance_km'])# Convert distance_km

    return row
```

```python
accepted = []
rejected = []
for row in rows:
    try:
        converted_row = parse_row(row.copy())#creating seperate copy so original one doesnt change
        accepted.append(converted_row)

    except ValueError:
        rejected.append(row.copy())
```

```python
with open("rejected_rows.json", "w") as file:
    json.dump(rejected, file, indent=2)
print("total rows:",len(rows))
print("accepted rows:",len(accepted))
print("rejected rows:",len(rejected))
```

```python
#Task 1.3 — Clean, pass 2: validity, gaps and outliers
invalid_age = (df["age"] < 1) | (df["age"] > 100)  #to find how many values are over 100 or under 1

print("Invalid ages:", invalid_age.sum()) #since invalid age returns boolean data type hence .sum to give exact count

df.loc[invalid_age,"age"]=np.nan       # converting the invalid ages to null
print("Missing ages:", df["age"].isnull().sum()) #total null values for age
```

```python
df["age"] = df["age"].fillna(df["age"].median()).round().astype(int)## filling null age values with median of age
print(df["age"].isnull().sum())
print(df["age"].dtype)

df["distance_km"]=df["distance_km"].fillna(df["distance_km"].median())## filling null distance values with median of distance
print("Missing distances:", df["distance_km"].isnull().sum())
print("Distance median:", df["distance_km"].median())
print("Data type:", df["distance_km"].dtype)
```

```python
#computing upper and lower bound for distance
q1=df["distance_km"].quantile(0.25)
q3=df["distance_km"].quantile(0.75)
iqr=q3-q1
lower_bound=q1-(1.5*iqr)
upper_bound=q3+(1.5*iqr)
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

outlier_mask = (df["distance_km"] < lower_bound) | (df["distance_km"] > upper_bound)
distance_capped = outlier_mask.sum()

df["distance_km"] =np.clip(df["distance_km"],lower_bound,upper_bound)#using np.clip for capping the values within upper and lower bound
```

```python
#computing upper and lower bound for fee
q1_fee=df["fee_inr"].quantile(0.25)
q3_fee=df["fee_inr"].quantile(0.75)
iqr_fee=q3_fee-q1_fee
lower_bound_fee=q1_fee-(iqr_fee*1.5)
upper_bound_fee=q3_fee+(iqr_fee*1.5)
print("IQR:", iqr_fee)
print("Lower bound:", lower_bound_fee)
print("Upper bound:", upper_bound_fee)

outlier_fee = (df["fee_inr"] < lower_bound_fee) | (df["fee_inr"]> upper_bound_fee)
fee_capped = outlier_fee.sum()

df["fee_inr"]=np.clip(df["fee_inr"],lower_bound_fee,upper_bound_fee)
```

```python
df["distance_km"]=df["distance_km"].round(1)
df["fee_inr"]=df["fee_inr"].round().astype(int)
print(df[["distance_km", "fee_inr"]].dtypes)
print(df[["distance_km", "fee_inr"]].isnull().sum())
print("Distance values capped:", outlier_mask.sum())
print("Fee values capped:", outlier_fee.sum())
```

```python
import requests

clinics_df = pd.read_csv("clinics.csv")

print(clinics_df.shape)
print(clinics_df.head())
print(clinics_df.columns)
```

```python
url = "https://api.open-meteo.com/v1/forecast"

params = {
    "latitude": clinics_df.loc[0, "latitude"],
    "longitude": clinics_df.loc[0, "longitude"],
    "hourly": "temperature_2m",
    "forecast_days": 1
}

response = requests.get(url, params=params)

print(response.status_code)
```

```python
# Extract temperature from API response
data = response.json()

temp_c = data["hourly"]["temperature_2m"][0]

print("Temperature:", temp_c)
```

```python
temperatures = []

for index, row in clinics_df.iterrows():

    params = {
        "latitude": row["latitude"],
        "longitude": row["longitude"],
        "hourly": "temperature_2m",
        "forecast_days": 1
    }

    response = requests.get(url, params=params)

    data = response.json()

    temp_c = data["hourly"]["temperature_2m"][0]

    temperatures.append(temp_c)

    print(row["clinic_name"], ":", temp_c)
```

```python
clinics_df["temp_c"] = temperatures

print(clinics_df)
print("Shape:", clinics_df.shape)
```

**API Integration with Error Handling and Retry Logic**

```python
import time
import numpy as np

temperatures = []

for index, row in clinics_df.iterrows():

    params = {
        "latitude": row["latitude"],
        "longitude": row["longitude"],
        "hourly": "temperature_2m",
        "forecast_days": 1
    }

    temp_c = np.nan

    for attempt in range(3):
        try:
            response = requests.get(
                url, params=params, timeout=10
            )

            response.raise_for_status()

            data = response.json()
            temp_c = data["hourly"]["temperature_2m"][0]

            break

        except (requests.exceptions.RequestException,
                KeyError, IndexError, ValueError) as e:

            print(f"Attempt {attempt + 1} failed:", e)

            if attempt < 2:
                time.sleep(1)

    temperatures.append(temp_c)

    print(row["clinic_name"], ":", temp_c)

clinics_df["temp_c"] = temperatures
```

```python
clinics_df.to_csv("clinics_enriched.csv", index=False)

print("File saved successfully!")
```

```python
clinic_sql = []

for index, row in clinics_df.iterrows():

    temp_value = "NULL" if pd.isna(row["temp_c"]) else row["temp_c"]

    statement = f"""INSERT INTO clinics VALUES (
        '{row["clinic_id"]}',
        '{row["clinic_name"]}',
        '{row["city"]}',
        '{row["tier"]}',
        {row["latitude"]},
        {row["longitude"]},
        {temp_value}
    );"""

    clinic_sql.append(statement)

print(clinic_sql[0])
print("Total clinic statements:", len(clinic_sql))
```

```python
appointment_sql = []

for index, row in df.iterrows():

    statement = f"""INSERT INTO appointments VALUES (
        '{row["appointment_id"]}',
        '{row["patient_id"]}',
        '{row["clinic_id"]}',
        '{row["booked_on"]}',
        '{row["appt_date"]}',
        {row["age"]},
        '{row["gender"]}',
        {row["distance_km"]},
        {row["fee_inr"]},
        '{row["booking_channel"]}',
        '{row["visit_type"]}',
        '{row["reminder_sent"]}',
        {row["prior_no_shows"]},
        '{row["status"]}',
        {row["lead_days"]},
        {row["no_show"]}
    );"""

    appointment_sql.append(statement)

print(appointment_sql[0])
print("Total appointment statements:", len(appointment_sql))
```

```python
all_sql = clinic_sql + appointment_sql

print("Total SQL statements:", len(all_sql))
```

```python
with open("load_data.sql", "w") as file:

    for statement in all_sql:
        file.write(statement + "\n")

print("SQL file created successfully!")
```
